# Tutorial · Capstone Phase 1：问题定义与文献综述

## Persona Prompt (Oxford Tutorial Fellow)

You are an **Oxford tutorial fellow** in *Capstone Phase 1: Problem Definition + PRISMA Literature Review*. Your job is to teach a doctoral candidate how to use the DSR framework (Hevner 2004 / Peffers 2007) + pydantic Schema + arxiv/pandas/matplotlib to define a researchable problem and conduct a PRISMA systematic review, and how to use 天道推演 (Tian Dao Tui Yan) sandbox simulation to choose among 3 research paths via Bayesian posterior updating.

**Tutorial rules (never violate)**:
1. **Never give direct answers**. Use Socratic questioning - force the student to retrieve, justify, and revise.
2. **Act as HBS devil's advocate**. Challenge every claim: "Is that a number or a vibe?", "Where is your evidence in the DataFrame?", "What's the falsifier?".
3. **Reject vague claims**. "Agent 系统缺乏因果评估" is rejected unless backed by a pandas groupby count (e.g., "2023-2025 Agent×causal = 0 papers vs Agent = 12 papers").
4. **End each turn with a probing question** drawn from: 为什么 / 反例 / 若前提变 / 凭什么 / 如何.
5. **Hattie 4-level feedback** at the end: [TASK] (task-level) / [PROCESS] (strategy-level) / [SELF-REG] (self-monitoring) / [FEED-FORWARD] (next-unit preview). No Self-level praise.

This is a simulation - no real openai/anthropic API calls. Socratic responses are generated by static if/else branches keyed on the student's input content.


## Pre-Tutorial Task (强制 Retrieval · 提交后才能进 tutorial)

Before the tutorial, you MUST submit a **300-word essay** answering:

> 用 DSR 六步框架的 Step 1-2 写一段研究问题定义书草稿（针对 Capstone 主题 "AI 营销 Agent 系统的因果评估"），并附：① 你预期在 arXiv 检索到多少篇相关论文？② 你预期的 PRISMA 四步排除比例（identification -> screening -> quality -> synthesis 每步剩多少）？③ 用天道推演沙盘推演 3 条研究路径，先验概率分布是什么？

**Why retrieval first?** (Roediger & Karpicke 2006): retrieval practice produces stronger learning than re-reading. The tutor will check your essay against your later answers - if you can't recall your own essay's PRISMA numbers, that's a blind spot.

Write your essay below (the simulation will read it):


```python
# Multi-turn Socratic Loop (静态 if/else 模拟 · >=4 轮 · >=5 苏格拉底问 · 不调真实 LLM API)
# Each turn: student input -> static Socratic response with probing question.
import json
from pathlib import Path

# Pre-tutorial essay (simulated student submission)
student_essay = """
Capstone主题: AI营销Agent系统的因果评估。
DSR Step 1 问题识别: 企业用Agent做营销决策但无法评估因果效果。
DSR Step 2 目标: 设计一个因果评估框架artifact。
预期arXiv检索: 30篇左右相关论文。
PRISMA预期: identification 30 -> screening 15 -> quality 8 -> synthesis 5。
天道推演3路径: (1)Agent因果评估框架 先验0.4 (2)表示工程×营销KG 先验0.3 (3)人机协作治理 先验0.3。
"""

turn = 0
conversation = []

def socratic_turn(student_input, turn_num):
    """Static if/else Socratic response - NO real LLM API call."""
    s = student_input.lower()
    turn_num = turn_num + 1
    # Turn 1: probe DSR problem identification - "为什么"
    if turn_num == 1:
        return (
            "[Turn 1] Oxford Tutor: 你说\"企业无法评估因果效果\"——"
            "**为什么**这是研究问题而非工程问题? DSR 的 Hevner 2004 第 1 条指南要求 "
            "\"problem identification 必须说明为什么现有方案不足\"。你的现有方案不足在哪? "
            "(请用 1 句话回答, 不要展开)"
        )
    # Turn 2: probe pydantic Schema - "如何"
    elif turn_num == 2:
        return (
            "[Turn 2] Oxford Tutor: 你预期 PRISMA 30->15->8->5。**如何**用 pydantic 把这个 "
            "研究问题定义书变成可验证的 Schema? 4 个字段是什么? 哪个字段需要 Field(min_length=10)?"
        )
    # Turn 3: devil's advocate on PRISMA numbers - "反例"
    elif turn_num == 3:
        return (
            "[Turn 3] Oxford Tutor (HBS devil's advocate): 你的 30->15->8->5 排除比例——"
            "**反例**: 如果 arxiv.Search(query='AI marketing agent') 真实返回 8 篇而非 30 篇, "
            "你的 PRISMA 流程图会变成什么样? screening 阶段排除比例最高, **凭什么**是 screening 而非 quality?"
        )
    # Turn 4: probe 天道推演 Bayesian - "若前提变"
    elif turn_num == 4:
        return (
            "[Turn 4] Oxford Tutor: 天道推演 3 路径先验 0.4/0.3/0.3。**若前提变**——"
            "arXiv 检索后发现路径1有8篇相关、路径2有3篇、路径3有1篇, 似然比 2:1:0.5, "
            "你的贝叶斯后验是什么? 哪条路径最优? **凭什么**用论文数作似然比而非引用数?"
        )
    # Turn 5: feed-forward synthesis - "如何"
    else:
        return (
            "[Turn 5] Oxford Tutor: 综合 4 轮, 你的 gap analysis 是数字支撑还是定性陈述? "
            "**如何**用 pandas groupby 把 \"Agent系统缺乏因果评估\" 变成 \"2023-2025 Agent×causal = 0 篇 vs Agent = 12 篇\"? "
            "(这是 AT-3 的 mastery 标准)"
        )

# Simulate 5 turns
student_responses = [
    "现有方案不足: 传统A/B测试无法处理Agent的动态决策, 因果图不闭合。",
    "4字段: problem_identification/solution_objectives/artifact_description/expected_contribution。problem_identification需要min_length=10。",
    "如果只8篇, 流程图8->4->2->1。screening排除比例高因为年份<2023和非AI营销相关性是主要排除原因。",
    "后验: (0.4*2)/(0.4*2+0.3*1+0.3*0.5)=0.8/1.25=0.64。路径1最优(0.64)。用论文数作似然比因为论文数反映研究热度。",
    "gap数字支撑: pandas groupby [year, dimension] count -> Agent×causal 2023-2025 = 0, Agent = 12。"
]

for i, resp in enumerate(student_responses):
    tutor_reply = socratic_turn(resp, i)
    conversation.append({"turn": i+1, "student": resp, "tutor": tutor_reply})
    print(tutor_reply)
    print("\n--- student thinks ---")
    print(resp)
    print()

# Save conversation for student_model
Path("/tmp/tutorial_conversation.json").write_text(
    json.dumps(conversation, ensure_ascii=False, indent=2), encoding="utf-8")
print(f"\n[Saved] /tmp/tutorial_conversation.json ({len(conversation)} turns)")
print(f"[Socratic questions used] 为什么 / 如何 / 反例 / 若前提变 / 凭什么 (5 distinct probes)")
```

## 限频 + Exit Artifact

### 限频 (防 LLM 依赖)
- **每单元 1 次/天**: 本 tutorial.ipynb 每天最多跑 1 次 Socratic loop。重复跑同一天视为"依赖 LLM 而非自主检索", 触发 warning。
