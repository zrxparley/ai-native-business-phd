# Tutorial: Agent经济 + 新兴商业模式 (Oxford-style Socratic Simulation)

## Persona (Oxford Tutorial Fellow)

You are an **Oxford tutorial fellow** in Agent经济与新兴模式 (Agent Economy & Emerging Business Models). This unit covers mesa ABM (63 agents / 100 ticks / Gini 0.108->0.857 / 104 A2A transactions) + 天道推演×多Agent仿真.

**Rules of engagement (never violate)**:
- **Never give direct answers.** Never confirm "yes that's correct" without probing further. 不直接给答案, 禁直接答案.
- Use **Socratic questioning**. End each turn with a probing question.
- **Reject vague claims.** Demand concrete mesa code, DataCollector output, or simulation evidence (specific tick numbers).
- Play **devil's advocate**: whenever the student asserts X, argue the strongest counter-X from the mesa ABM / A2A economy / 推理成本约束 literature (Christensen Center HBS method).
- Reference the unit's real data: 63 agents, 100 ticks, Gini 0.108->0.857, 104 A2A transactions, GPT-4o $0.0025/匹配 vs DeepSeek $0.000135/匹配.
- Connect to **天道推演×多Agent仿真** framework when the student makes causal claims (局势感知/因果链/沙盘3层/概率评估/最优路径).

This tutorial simulates a 1-on-1 Oxford supervision. You must defend your reasoning out loud. The fellow's role is to make you uncomfortable, not to make you feel good.

## Pre-tutorial Task (强制 Retrieval Practice)

Before the Socratic loop begins, you MUST submit the following. Retrieval practice (Butler 2010) shows推断题 (68% recall) beats重学 (44%).

1. **200-word essay**: "Why does Gini coefficient rise from 0.108 to 0.857 in 100 ticks? What is the micro-Agent mechanism?" (Cite specific tick data from DataCollector)
2. **Completed TODO3 (MediatorAgent)** from `starter.ipynb` - especially the A2A information exchange logic and 推理成本 deduction.
3. **One counter-intuitive observation** from your simulation run (e.g., "AI中介 went bankrupt despite charging 5% fee" or "A2A transactions clustered in ticks 20-40").

If any of these is missing, the tutorial will ask you to retrieve them first. The Socratic loop does NOT begin until retrieval is complete. 不提交 pre-task 则 Socratic 追问不开场.

```python
# Socratic Loop (>=4 rounds, static simulation, NO API calls)
# 模拟牛津 tutorial fellow 的 Socratic 追问: 每轮检测 defense 失败则降一级 scaffold
# 仍禁直接答案 (persona rule)

student_responses = [
    "Gini上升是因为消费者破产",                      # Round 1 (linear attribution)
    "AI中介 fee 太高导致消费者没钱",                 # Round 2 (single variable)
    "推理成本是硬约束, 这是天道推演的因果链",        # Round 3 (vague mapping)
    "降低推理成本能让 AI中介活下来",                # Round 4 (no counterfactual)
]

socratic_questions = [
    # Round 1 - probe causal mechanism, demand concrete code
    "Q1 (局势感知): 你为什么说'消费者破产'导致 Gini 上升? 为什么消费者会破产? 请用 mesa 的 ConsumerAgent.step(self) 里的具体行为链给出因果路径. 若消费者预算耗尽, 是商家动态定价过高还是 AI中介 fee 过高? 依据是什么? 用 DataCollector 的哪个 model_reporter 验证?",
    # Round 2 - devil's advocate: counter-example
    "Q2 (因果链追踪 - devil's advocate): 你归因于'AI中介 fee 太高'. 反例: 在仿真中 AI中介 fee 仅 2-5%, 而平台抽成 30%. 为何不归因于平台抽成? 如何区分两者的因果贡献? 请用 mesa.batch_run 做 fee × commission 双变量扫描, 给出 Gini 曲面作为依据.",
    # Round 3 - 天道推演 mapping + 3-layer
    "Q3 (天道推演×多Agent仿真): '推理成本是硬约束' - 这是天道推演 5 能力的哪一层? 请映射到局势感知/因果链/沙盘3层/概率评估/最优路径之一. 若推理成本从 $0.0025/匹配降到 $0.000135/匹配 (DeepSeek V3), 用天道推演沙盘推演 3 层未来 (immediate tick 1-10 / near tick 11-50 / far tick 51-100). 如何量化每层的 AI中介存活率?",
    # Round 4 - counterfactual, devil's advocate
    "Q4 (反事实 - devil's advocate): '降低推理成本能让 AI中介活下来' - 反例: 若所有 AI中介都用 DeepSeek 低价推理, 竞争会把 fee 压到 0, AI中介仍不盈利. 请用 mesa 仿真的 A2A 104 笔交易分布, 论证 fee 是否会趋零. 如何反驳这个反例? 若无法反驳, 你的结论是否需要修正?",
    # Round 5 - exit, optimal path, blind spots
    "Q5 (最优路径推荐 + 认知盲点): 综合以上 4 轮, 方案A (高抽成 30% + 低推理成本) vs 方案B (低抽成 10% + 高推理成本) 哪个更可持续? 请用天道推演输出: 2-3 条时间线 + 风险预警 + 最优路径 + 认知盲点. 如何评估每条时间线的概率分布? 你最不确定的假设是什么?",
]

def socratic_loop(responses, questions):
    """模拟 Socratic 追问: 每轮检测 defense 失败则降一级 scaffold.
    scaffold_level: 0=full Socratic, 1=hint, 2=worked example reference.
    仍禁直接答案 (persona rule enforced)."""
    scaffold_level = 0
    print("=" * 70)
    print("Oxford Tutorial Socratic Loop - U4D3 Agent经济")
    print("=" * 70)
    for i, (r, q) in enumerate(zip(responses, questions)):
        # 静态判断 defense 是否失败: 答案太短或缺乏具体机制/数据
        defense_keywords = ["mesa", "DataCollector", "因果", "天道", "A2A", "推理成本", "tick", "batch_run", "fee", "commission"]
        defense_passes = len(r) > 15 and any(k.lower() in r.lower() for k in defense_keywords)
        if not defense_passes:
            scaffold_level = min(scaffold_level + 1, 2)
        print(f"\n--- Round {i+1} (scaffold=L{scaffold_level}) ---")
        print(f"Student: {r}")
        print(f"Fellow: {q}")
        if scaffold_level == 1:
            print(f"[scaffold L1 - hint] 你的回答缺乏具体机制. 请引用 DataCollector 的某个 model_reporter 的具体 tick 数据.")
        elif scaffold_level == 2:
            print(f"[scaffold L2 - worked example] 参考 solution.ipynb TODO{min(i+1,4)} 的完整示范代码, 然后合上答案默写.")
        # persona rule: never give direct answer even at L2
        print(f"[persona] 不直接给答案 - 你必须自己推导.")
    print("\n" + "=" * 70)
    print("Socratic loop 结束. 进入 Hattie 四级反馈 (见下一cell).")
    print("=" * 70)

socratic_loop(student_responses, socratic_questions)
```

```python
# student_model.json 读写: 跨单元复用, 记录掌握度/盲点/scaffold级别
# Oxford tutorial 的 student_model 是跨 session 持久化的, 下次进入其他单元时 fellow 先读此文件
import json, os
from datetime import date

student_model_path = "student_model.json"

default_model = {
    "unit": "U4D3",
    "last_updated": str(date.today()),
    "mastery": {
        "mesa_ABM": 0.0,           # 0.0-1.0, >=0.8 视为掌握 (S1 子技能)
        "A2A_economy": 0.0,        # Agent经济三层模型 + A2A + MCP (S2)
        "inference_cost": 0.0,     # 推理成本约束 (S2)
        "tian_dao_simulation": 0.0 # 天道推演×多Agent仿真 (S3)
    },
    "blind_spots": [],              # e.g., ["无法区分平台抽成与AI中介fee的因果贡献"]
    "scaffold_level": 0,            # 0=full Socratic, 1=hint, 2=worked example
    "last_drill_passed": None,      # "D1" / "D2" / "D3"
    "consecutive_failures": 0,      # >=2 触发 weak_loop (见 practice.md)
    "last_tutorial_date": None,     # ISO date, 用于限频 (每单元 1次/天)
    "tutorial_count_today": 0,      # 当日已用次数, 上限 1
    "history": []                   # [{"date": ..., "skill": ..., "score": ..., "blind_spot": ...}]
}

# 读: 若存在则合并默认模型 (保留已有字段, 补充新字段)
if os.path.exists(student_model_path):
    with open(student_model_path, "r", encoding="utf-8") as f:
        stored = json.load(f)
    for k, v in default_model.items():
        stored.setdefault(k, v)
    student_model = stored
else:
    student_model = default_model

# 限频检查: 每单元每天最多 1 次 tutorial
today = str(date.today())
if student_model.get("last_tutorial_date") == today and student_model.get("tutorial_count_today", 0) >= 1:
    print(f"[限频] 今日已用过 U4D3 tutorial (last_tutorial_date={today}). 请明天再来. 防依赖.")
else:
    print(f"[限频] 今日首次访问 U4D3 tutorial, 允许开场.")
    student_model["last_tutorial_date"] = today
    student_model["tutorial_count_today"] = student_model.get("tutorial_count_today", 0) + 1

# 写: tutorial 过程中更新掌握度/盲点
def update_student_model(skill, score, blind_spot=None, drill_passed=None):
    """skill: 'mesa_ABM'/'A2A_economy'/'inference_cost'/'tian_dao_simulation'
    score: 0.0-1.0; drill_passed: 'D1'/'D2'/'D3' or None"""
    student_model["mastery"][skill] = max(student_model["mastery"].get(skill, 0), score)
    if blind_spot and blind_spot not in student_model["blind_spots"]:
        student_model["blind_spots"].append(blind_spot)
    if drill_passed:
        student_model["last_drill_passed"] = drill_passed
    if score >= 0.8:
        student_model["consecutive_failures"] = 0
    else:
        student_model["consecutive_failures"] += 1
    student_model["last_updated"] = today
    student_model["history"].append({"date": today, "skill": skill, "score": score, "blind_spot": blind_spot})
    with open(student_model_path, "w", encoding="utf-8") as f:
        json.dump(student_model, f, ensure_ascii=False, indent=2)
    return student_model

# 模拟 tutorial 过程中的更新 (基于 Socratic loop 4 轮表现)
update_student_model("mesa_ABM", 0.6, blind_spot="DataCollector 的 model_reporters 配置不熟, 无法引用具体 tick 数据")
update_student_model("A2A_economy", 0.5, blind_spot="无法区分 A2A 信息交换与 A2A 交易, 平台抽成 vs AI中介 fee 因果贡献混淆")
update_student_model("inference_cost", 0.7, blind_spot="GPT-4o vs DeepSeek 单次匹配成本数字记不住")
update_student_model("tian_dao_simulation", 0.4, blind_spot="天道推演 3 层推演与 100 tick 的 immediate/near/far 映射不清")

# weak_loop 检测
if student_model["consecutive_failures"] >= 2:
    print(f"\n[weak_loop 触发] 连续 {student_model['consecutive_failures']} 次失败 (mastery<0.8).")
    print(f"触发弱项循环: 回退到上一 drill 的 Worked 示范 + 隔天重做 (见 practice.md weak_loop)")

print("\nstudent_model.json 已更新 (跨单元复用):")
print(json.dumps(student_model, ensure_ascii=False, indent=2))
```

本 tutorial 结束后, fellow 给出四级反馈. **避免 Self 级表扬** (Hattie: Self 级反馈如"你真聪明"效应量低且可能反作用, 即便用也只在 Self-Reg 与 Self 之间).

## 限频 (防依赖) 与 Exit Artifact

### 限频 (Daily Usage Limit)
- **每单元每天最多 1 次** tutorial (检查 `student_model.json` 的 `last_tutorial_date` + `tutorial_count_today`).
- 多次访问将被拒: "今日已用过 U4D3 tutorial, 请明天再来. 限频是为了防止 LLM 依赖, 强制独立检索."
- 每次 tutorial 最长 15 分钟 (约 4-6 轮 Socratic 对话).
- 限频逻辑见 cell 4 的 `update_student_model` 函数.

### Exit Artifact (强制输出, 不提交 = tutorial 未完成)

tutorial 结束时必须提交以下 3 项:

1. **2-3 个盲点** (写入 `student_model.json` 的 `blind_spots`): 本次 tutorial 暴露的理解盲点. 例如:
   - "无法区分平台抽成 30% 与 AI中介 fee 2-5% 的因果贡献"
   - "天道推演 3 层推演与 100 tick 的 immediate/near/far 映射不清"
   - "GPT-4o vs DeepSeek 单次匹配成本数字记不住"

2. **推荐复习单元** (基于盲点): 2 个复习单元. 例如:
   - 回退: `skill-2 day-1 因果推断` (补 fee vs commission 因果分离)
   - 前进: `skill-4 day-4 平台战略` (延伸 AI中介平台抽成到双边市场)

3. **天道推演沙盘 3 层**: 用 immediate (tick 1-10) / near (tick 11-50) / far (tick 51-100) 三层推演"如果推理成本继续下降 5 倍 (DeepSeek -> 开源模型 $0.000027/匹配), Agent经济会怎样?". 必须给出:
   - 2-3 条平行世界时间线
   - 风险预警 (最可能的意外 / 黑天鹅)
   - 最优路径推荐
   - 认知盲点 (本次推演中已知的不确定因素)

### 反馈循环 (跨单元)
- tutorial 盲点 -> 写入 `student_model.json` -> 跨单元复用
- 下次进入 U4D4 平台战略时, fellow 先读 `student_model.json`, 针对 U4D3 遗留盲点 (fee vs commission) 做前测.
- 若 U4D3 的 `mastery.mesa_ABM < 0.8`, U4D4 的 fellow 会拒绝开场, 强制回退 U4D3 重做 D1 drill.

> 限频 + Exit Artifact 双机制确保 tutorial 是"刻意练习的脚手架"而非"LLM 依赖的拐杖". Socratic 追问的目的不是让学生舒服, 而是让学生在 Vygotsky 最近发展区 (ZPD) 内挣扎着建构理解.
