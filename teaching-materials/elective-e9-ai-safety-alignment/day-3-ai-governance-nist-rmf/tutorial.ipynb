## Hattie 4 级 Formative Feedback (避免 Self 级表扬)

> Hattie 可见学习: 反馈应在 Task / Process / Self-Reg / Feed-Forward 4 级, 避免 Self 级 (表扬/批评个人, 无助学习)

### [TASK] 任务级反馈 (针对本次 essay 的具体内容)
- **对**: 你正确识别了 NIST 四步 (Govern/Map/Measure/Manage) 和 EU AI Act 4 级框架
- **错**: "Govern 是第一步" - 误读。NIST 官方明确 Govern **贯穿**全过程, 不是线性第一步。你的 pydantic schema 若把 Govern 写成线性 step1 而非 cross-cutting, 会遗漏 GOVERN-2 在 Manage 阶段的问责回溯
- **修**: 重写 GOVERN-2 (问责结构) 的 assess_control, 加入"Manage 阶段风险事件触发 Govern 层问责"的因果链

### [PROCESS] 过程级反馈 (针对学生的推理策略)
- **对**: 你用"维度不同"解释三框架互补, 策略正确 (NIST=方法论 / EU=法律分级 / 中国=备案标识)
- **错**: 推理停留在概念层, 未用真实条款号举证。凭什么说 AI 动态定价在 EU 是"最小风险"? 应引用 Annex III 第几条 (保险定价升级路径)
- **修**: 每个风险分级判断必须附 NIST 控制项编号 OR EU 条款号 OR 中国法规名, 否则视为"vague claim"被驳回

### [SELF-REG] 自我调节级反馈 (针对学生的元认知)
- **对**: 你在 Turn 4 主动识别"computer use 需要 UI 级紧急停止, 不能只靠 API 级 MCP" - 显示自我纠错能力
- **错**: 你未主动检测"garak 通过不等于安全"的逻辑漏洞 - 这是证实偏差 (confirmation bias)。红队测试是证伪工具, 不能证实安全
